````python
import bchlib
import numpy as np
import matplotlib.pyplot as plt

# 1. Define BCH Parameters (n, k, t)
n = 255  # Codeword length (Container)
k = 131  # Message length (Secret Key) - chosen to allow for some error correction
t = 13   # Error correction capability (Wiggle Room) - Adjusted to be valid for n=255, k=131

# Initialize BCH object
bch = bchlib.BCH(m=8, t=t)

# 2. Simulate Key Generation
original_data_bytes = np.random.randint(0, 256, size=(k + 7) // 8, dtype=np.uint8)

# 3. Encode the Key
parity = bch.encode(original_data_bytes)
# Explicitly convert to bytearray to avoid UFuncTypeError
codeword = bytearray(original_data_bytes) + bytearray(parity)

# 4. Simulate Noise (Heart Rate Noise)
def introduce_errors(codeword_bytes, num_errors):
    noisy_codeword = bytearray(codeword_bytes)
    codeword_bits = len(noisy_codeword) * 8

    error_positions = np.random.choice(codeword_bits, num_errors, replace=False)

    for pos in error_positions:
        byte_idx = pos // 8
        bit_idx = pos % 8
        noisy_codeword[byte_idx] ^= (1 << bit_idx)
    return noisy_codeword

# 5. Decode the Message and Demonstrate Recovery
max_errors_to_test = t * 2
success_rates = []
num_trials = 100

print(f"Original Data (bytes): {list(original_data_bytes)}")

for num_errors in range(max_errors_to_test + 1):
    successful_recoveries = 0
    for _ in range(num_trials):
        noisy_codeword = introduce_errors(codeword, num_errors)

        parity_len = len(parity)
        noisy_data = noisy_codeword[:-parity_len]
        noisy_parity = noisy_codeword[-parity_len:]

        # Handle both success (tuple) and failure (int -1) return types
        decode_result = bch.decode(noisy_data, noisy_parity)
        if isinstance(decode_result, tuple):
            if len(decode_result) == 3:
                num_corrected_errors, decoded_data_bytes, _ = decode_result
            elif len(decode_result) == 2:
                num_corrected_errors, decoded_data_bytes = decode_result
            else:
                num_corrected_errors, decoded_data_bytes = -1, noisy_data
        else:
            num_corrected_errors = decode_result
            decoded_data_bytes = noisy_data # Decoding failed, original noisy data remains

        if num_corrected_errors >= 0 and np.array_equal(original_data_bytes, decoded_data_bytes[:len(original_data_bytes)]):
            successful_recoveries += 1

    success_rate = successful_recoveries / num_trials
    success_rates.append(success_rate)
    print(f"Errors introduced: {num_errors}, Success Rate: {success_rate:.2f}")

# Plotting the results
fig = plt.figure(figsize=(10, 6))
plt.plot(range(max_errors_to_test + 1), success_rates, marker='o')
plt.axvline(x=t, color='r', linestyle='--', label=f'BCH Correction Capability (t={t})')
plt.title('BCH Code Error Correction Success Rate vs. Number of Errors')
plt.xlabel('Number of Bit Errors Introduced')
plt.ylabel('Success Rate of Data Recovery')
plt.grid(True)
plt.legend()
plt.xticks(np.arange(0, max_errors_to_test + 1, 5))
plt.yticks(np.arange(0, 1.1, 0.1))
plt.show()
````

### Explanation of the BCH Code Simulation:

This script models the Fuzzy Commitment Scheme using a Binary BCH (Bose-Chaudhuri-Hocquenghem) code, demonstrating its error-correction capabilities. Here's a breakdown:

1.  **BCH Parameters `(n, k, t)`**: We define `n` (codeword length), `k` (message length, representing the secret key), and `t` (error correction capability). For `n=255`, we derive `m=8` (extension degree of the Galois Field) to initialize the `bchlib`.
2.  **Key Generation**: A random byte array of `k` bits is generated to simulate the unique biometric 'secret key'.
3.  **Encoding**: The `bch.encode()` method takes the `original_data_bytes` and generates `parity` bytes. These parity bytes are appended to the original data to form the `codeword`.
4.  **Noise Simulation**: The `introduce_errors()` function simulates 'heart rate noise' by flipping a specified number of random bits in the `codeword`. This represents the natural fluctuations in a user's biometric signal.
5.  **Decoding and Recovery**: The `bch.decode()` method attempts to correct errors in the `noisy_codeword` using the `parity` information. It returns the number of errors corrected and the `decoded_data_bytes`.
6.  **Success Rate Calculation**: The simulation runs multiple trials (`num_trials`) for various numbers of introduced errors. It checks if the `decoded_data_bytes` match the `original_data_bytes` to calculate a success rate.
7.  **Visualization**: A plot illustrates how the success rate of recovering the original secret key changes as more errors (noise) are introduced. You'll observe that the success rate is 100% up to `t` errors, and then it drops significantly, demonstrating the `t` (wiggle room) property of the BCH code.

````python
solidity_code = """// SPDX-License-Identifier: MIT
pragma solidity ^0.8.19;

/**
 * @title PhoenixRulebook
 * @dev Immutable RBAC and Soul-Bound Token (SBT) registry for the Phoenix Sovereign Ecosystem.
 */
contract PhoenixRulebook {
    address public owner;

    // The core rulebook: appWallet => dataType => role => hasAccess
    mapping(address => mapping(bytes32 => mapping(bytes32 => bool))) private _permissions;

    // Tracks whether a wallet holds a valid Soul-Bound Token (SBT) identity
    mapping(address => bool) public hasSoulBoundToken;

    // Events for the n8n gatekeeper to index and listen to
    event PermissionGranted(address indexed appWallet, bytes32 indexed dataType, bytes32 indexed role);
    event PermissionRevoked(address indexed appWallet, bytes32 indexed dataType, bytes32 indexed role);
    event SBTMinted(address indexed appWallet);
    event SBTRevoked(address indexed appWallet);

    modifier onlyOwner() {
        require(msg.sender == owner, "Not authorized: Owner only");
        _;
    }

    constructor() {
        owner = msg.sender;
    }

    /**
     * @dev Mint a non-transferable SBT to a new app wallet and grant its first permission.
     */
    function onboardApp(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        require(!hasSoulBoundToken[appWallet], "App is already onboarded");

        // Mint SBT (Logical representation)
        hasSoulBoundToken[appWallet] = true;
        emit SBTMinted(appWallet);

        // Grant initial permission
        _permissions[appWallet][dataType][role] = true;
        emit PermissionGranted(appWallet, dataType, role);
    }

    /**
     * @dev Grant an additional permission to an already onboarded app.
     */
    function grantPermission(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        require(hasSoulBoundToken[appWallet], "App must have an SBT to receive permissions");

        _permissions[appWallet][dataType][role] = true;
        emit PermissionGranted(appWallet, dataType, role);
    }

    /**
     * @dev Granularly revoke a specific permission from an app.
     */
    function revokePermission(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        _permissions[appWallet][dataType][role] = false;
        emit PermissionRevoked(appWallet, dataType, role);
    }

    /**
     * @dev The "Nuke" button: Revokes the app's SBT, instantly voiding ALL permissions.
     */
    function nukeApp(address appWallet) external onlyOwner {
        require(hasSoulBoundToken[appWallet], "App does not have an active SBT");

        hasSoulBoundToken[appWallet] = false;
        emit SBTRevoked(appWallet);
    }

    /**
     * @dev Free gatekeeper check used by n8n. Costs no gas.
     * Evaluates if the app has an active SBT AND the specific data/role permission.
     */
    function hasPermission(address appWallet, bytes32 dataType, bytes32 role) external view returns (bool) {
        // If the SBT was revoked, all access is inherently denied regardless of granular mappings
        if (!hasSoulBoundToken[appWallet]) {
            return false;
        }
        return _permissions[appWallet][dataType][role];
    }
}
"""

# Save the smart contract to the local directory
filename = 'PhoenixRulebook.sol'
with open(filename, 'w') as f:
    f.write(solidity_code)

print(f"Successfully wrote the smart contract to {filename}.")
print("You can now download this file and deploy it via Remix IDE or Hardhat.")
````

### 1. The Code Node: Verifying the Signature
In your n8n **Code Node**, you will use this JavaScript to read the incoming webhook and verify the cryptographic signature against the claimed wallet:

```javascript
// Ensure n8n has NODE_FUNCTION_ALLOW_EXTERNAL=ethers in its environment variables
const ethers = require('ethers');

// 1. Extract data from the incoming webhook
const claimedWallet = $input.item.json.headers['x-app-wallet'];
const timestamp = $input.item.json.body.timestamp.toString();
const signature = $input.item.json.body.signature;

// 2. Prevent replay attacks (e.g., reject if older than 5 minutes)
const currentTime = Date.now();
if (currentTime - parseInt(timestamp) > 300000) {
    return { isAuthenticated: false, reason: "Timestamp expired" };
}

// 3. Verify Signature
let recoveredWallet;
try {
    recoveredWallet = ethers.verifyMessage(timestamp, signature);
} catch (error) {
    return { isAuthenticated: false, reason: "Invalid signature format" };
}

// 4. The Verdict
if (recoveredWallet.toLowerCase() === claimedWallet.toLowerCase()) {
    return { isAuthenticated: true, wallet: recoveredWallet };
} else {
    return { isAuthenticated: false, reason: "Wallet mismatch" };
}
```

````python
import bchlib
import numpy as np
import matplotlib.pyplot as plt

# 1. Define BCH Parameters (n, k, t)
n = 255  # Codeword length (Container)
k = 131  # Message length (Secret Key) - chosen to allow for some error correction
t = 13   # Error correction capability (Wiggle Room) - Adjusted to be valid for n=255, k=131

# Initialize BCH object
bch = bchlib.BCH(m=8, t=t)

# 2. Simulate Key Generation
original_data_bytes = np.random.randint(0, 256, size=(k + 7) // 8, dtype=np.uint8)

# 3. Encode the Key
parity = bch.encode(original_data_bytes)
# Explicitly convert to bytearray to avoid UFuncTypeError
codeword = bytearray(original_data_bytes) + bytearray(parity)

# 4. Simulate Noise (Heart Rate Noise)
def introduce_errors(codeword_bytes, num_errors):
    noisy_codeword = bytearray(codeword_bytes)
    codeword_bits = len(noisy_codeword) * 8

    error_positions = np.random.choice(codeword_bits, num_errors, replace=False)

    for pos in error_positions:
        byte_idx = pos // 8
        bit_idx = pos % 8
        noisy_codeword[byte_idx] ^= (1 << bit_idx)
    return noisy_codeword

# 5. Decode the Message and Demonstrate Recovery
max_errors_to_test = t * 2
success_rates = []
num_trials = 100

print(f"Original Data (bytes): {list(original_data_bytes)}")

for num_errors in range(max_errors_to_test + 1):
    successful_recoveries = 0
    for _ in range(num_trials):
        noisy_codeword = introduce_errors(codeword, num_errors)

        parity_len = len(parity)
        noisy_data = noisy_codeword[:-parity_len]
        noisy_parity = noisy_codeword[-parity_len:]

        # Handle both success (tuple) and failure (int -1) return types
        decode_result = bch.decode(noisy_data, noisy_parity)
        if isinstance(decode_result, tuple):
            if len(decode_result) == 3:
                num_corrected_errors, decoded_data_bytes, _ = decode_result
            elif len(decode_result) == 2:
                num_corrected_errors, decoded_data_bytes = decode_result
            else:
                num_corrected_errors, decoded_data_bytes = -1, noisy_data
        else:
            num_corrected_errors = decode_result
            decoded_data_bytes = noisy_data # Decoding failed, original noisy data remains

        if num_corrected_errors >= 0 and np.array_equal(original_data_bytes, decoded_data_bytes[:len(original_data_bytes)]):
            successful_recoveries += 1

    success_rate = successful_recoveries / num_trials
    success_rates.append(success_rate)
    print(f"Errors introduced: {num_errors}, Success Rate: {success_rate:.2f}")

# Plotting the results
fig = plt.figure(figsize=(10, 6))
plt.plot(range(max_errors_to_test + 1), success_rates, marker='o')
plt.axvline(x=t, color='r', linestyle='--', label=f'BCH Correction Capability (t={t})')
plt.title('BCH Code Error Correction Success Rate vs. Number of Errors')
plt.xlabel('Number of Bit Errors Introduced')
plt.ylabel('Success Rate of Data Recovery')
plt.grid(True)
plt.legend()
plt.xticks(np.arange(0, max_errors_to_test + 1, 5))
plt.yticks(np.arange(0, 1.1, 0.1))
plt.show()
````

### Explanation of the BCH Code Simulation:

This script models the Fuzzy Commitment Scheme using a Binary BCH (Bose-Chaudhuri-Hocquenghem) code, demonstrating its error-correction capabilities. Here's a breakdown:

1.  **BCH Parameters `(n, k, t)`**: We define `n` (codeword length), `k` (message length, representing the secret key), and `t` (error correction capability). For `n=255`, we derive `m=8` (extension degree of the Galois Field) to initialize the `bchlib`.
2.  **Key Generation**: A random byte array of `k` bits is generated to simulate the unique biometric 'secret key'.
3.  **Encoding**: The `bch.encode()` method takes the `original_data_bytes` and generates `parity` bytes. These parity bytes are appended to the original data to form the `codeword`.
4.  **Noise Simulation**: The `introduce_errors()` function simulates 'heart rate noise' by flipping a specified number of random bits in the `codeword`. This represents the natural fluctuations in a user's biometric signal.
5.  **Decoding and Recovery**: The `bch.decode()` method attempts to correct errors in the `noisy_codeword` using the `parity` information. It returns the number of errors corrected and the `decoded_data_bytes`.
6.  **Success Rate Calculation**: The simulation runs multiple trials (`num_trials`) for various numbers of introduced errors. It checks if the `decoded_data_bytes` match the `original_data_bytes` to calculate a success rate.
7.  **Visualization**: A plot illustrates how the success rate of recovering the original secret key changes as more errors (noise) are introduced. You'll observe that the success rate is 100% up to `t` errors, and then it drops significantly, demonstrating the `t` (wiggle room) property of the BCH code.

````python
solidity_code = """// SPDX-License-Identifier: MIT
pragma solidity ^0.8.19;

/**
 * @title PhoenixRulebook
 * @dev Immutable RBAC and Soul-Bound Token (SBT) registry for the Phoenix Sovereign Ecosystem.
 */
contract PhoenixRulebook {
    address public owner;

    // The core rulebook: appWallet => dataType => role => hasAccess
    mapping(address => mapping(bytes32 => mapping(bytes32 => bool))) private _permissions;

    // Tracks whether a wallet holds a valid Soul-Bound Token (SBT) identity
    mapping(address => bool) public hasSoulBoundToken;

    // Events for the n8n gatekeeper to index and listen to
    event PermissionGranted(address indexed appWallet, bytes32 indexed dataType, bytes32 indexed role);
    event PermissionRevoked(address indexed appWallet, bytes32 indexed dataType, bytes32 indexed role);
    event SBTMinted(address indexed appWallet);
    event SBTRevoked(address indexed appWallet);

    modifier onlyOwner() {
        require(msg.sender == owner, "Not authorized: Owner only");
        _;
    }

    constructor() {
        owner = msg.sender;
    }

    /**
     * @dev Mint a non-transferable SBT to a new app wallet and grant its first permission.
     */
    function onboardApp(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        require(!hasSoulBoundToken[appWallet], "App is already onboarded");

        // Mint SBT (Logical representation)
        hasSoulBoundToken[appWallet] = true;
        emit SBTMinted(appWallet);

        // Grant initial permission
        _permissions[appWallet][dataType][role] = true;
        emit PermissionGranted(appWallet, dataType, role);
    }

    /**
     * @dev Grant an additional permission to an already onboarded app.
     */
    function grantPermission(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        require(hasSoulBoundToken[appWallet], "App must have an SBT to receive permissions");

        _permissions[appWallet][dataType][role] = true;
        emit PermissionGranted(appWallet, dataType, role);
    }

    /**
     * @dev Granularly revoke a specific permission from an app.
     */
    function revokePermission(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        _permissions[appWallet][dataType][role] = false;
        emit PermissionRevoked(appWallet, dataType, role);
    }

    /**
     * @dev The "Nuke" button: Revokes the app's SBT, instantly voiding ALL permissions.
     */
    function nukeApp(address appWallet) external onlyOwner {
        require(hasSoulBoundToken[appWallet], "App does not have an active SBT");

        hasSoulBoundToken[appWallet] = false;
        emit SBTRevoked(appWallet);
    }

    /**
     * @dev Free gatekeeper check used by n8n. Costs no gas.
     * Evaluates if the app has an active SBT AND the specific data/role permission.
     */
    function hasPermission(address appWallet, bytes32 dataType, bytes32 role) external view returns (bool) {
        // If the SBT was revoked, all access is inherently denied regardless of granular mappings
        if (!hasSoulBoundToken[appWallet]) {
            return false;
        }
        return _permissions[appWallet][dataType][role];
    }
}
"""

# Save the smart contract to the local directory
filename = 'PhoenixRulebook.sol'
with open(filename, 'w') as f:
    f.write(solidity_code)

print(f"Successfully wrote the smart contract to {filename}.")
print("You can now download this file and deploy it via Remix IDE or Hardhat.")
````

## Energy Savings Calculation for Spanish Courtyard Tiny Home

````python
import math

def calculate_golden_ratio_courtyard(base_dimension_m: float) -> dict:
    """
    Calculates courtyard dimensions based on the Golden Ratio, ensuring both
    dimensions are multiples of 3.

    Args:
        base_dimension_m (float): The initial base dimension (e.g., width or length) in meters.

    Returns:
        dict: A dictionary containing 'short_side_m', 'long_side_m', and 'phi_value'.
    """
    PHI = (1 + math.sqrt(5)) / 2 # Golden Ratio

    # Ensure the base_dimension is a multiple of 3
    # Round to the nearest multiple of 3
    short_side_m = round(base_dimension_m / 3) * 3
    if short_side_m == 0: # Ensure minimum dimension if base_dimension was too small
        short_side_m = 3

    # Calculate the long side using the golden ratio
    raw_long_side_m = short_side_m * PHI

    # Adjust the long side to be the nearest multiple of 3
    long_side_m = round(raw_long_side_m / 3) * 3
    if long_side_m == 0: # Ensure minimum dimension if calculated was too small
        long_side_m = 3

    return {
        "short_side_m": short_side_m,
        "long_side_m": long_side_m,
        "phi_value": PHI
    }

# Demonstrate the function
print("--- Golden Ratio Courtyard Dimensions ---")

# Example 1: Starting with a base dimension of 6.5 meters
base_dim_1 = 6.5
courtyard_dims_1 = calculate_golden_ratio_courtyard(base_dim_1)
print(f"Base Dimension: {base_dim_1:.2f} m")
print(f"  Short Side: {courtyard_dims_1['short_side_m']:.2f} m (Multiple of 3)")
print(f"  Long Side: {courtyard_dims_1['long_side_m']:.2f} m (Multiple of 3)")
print(f"  Golden Ratio (Phi): {courtyard_dims_1['phi_value']:.4f}\n")

# Example 2: Starting with a base dimension of 10 meters
base_dim_2 = 10.0
courtyard_dims_2 = calculate_golden_ratio_courtyard(base_dim_2)
print(f"Base Dimension: {base_dim_2:.2f} m")
print(f"  Short Side: {courtyard_dims_2['short_side_m']:.2f} m (Multiple of 3)")
print(f"  Long Side: {courtyard_dims_2['long_side_m']:.2f} m (Multiple of 3)")
print(f"  Golden Ratio (Phi): {courtyard_dims_2['phi_value']:.4f}")

# Example 3: Starting with a base dimension of 2.1 meters (demonstrates rounding up to nearest multiple of 3)
base_dim_3 = 2.1
courtyard_dims_3 = calculate_golden_ratio_courtyard(base_dim_3)
print(f"Base Dimension: {base_dim_3:.2f} m")
print(f"  Short Side: {courtyard_dims_3['short_side_m']:.2f} m (Multiple of 3)")
print(f"  Long Side: {courtyard_dims_3['long_side_m']:.2f} m (Multiple of 3)")
print(f"  Golden Ratio (Phi): {courtyard_dims_3['phi_value']:.4f}")
````

### Python Simulation: Warden `Δ_frame` Monitoring & Escalation

This script demonstrates the active control architecture described in the IP addendum. Rather than just logging data, the Warden actively measures **semantic drift** (`Δ_frame`) introduced by its own selection frame.

If the selected candidate drifts too far from the raw observation's true semantic density, the system explicitly **halts/escalates** rather than passing hallucinated or drifted context to the next sub-system.

````python
import json
from datetime import datetime, timezone

class WardenActiveControl:
    def __init__(self, max_allowable_drift: float):
        # The maximum allowed semantic deviation (Δ_frame) before triggering an escalation
        self.max_allowable_drift = max_allowable_drift
        self.truncation_depth = 1 # k-depth boundary

    def process_observation(self, observation_id: str, candidate_space: list) -> dict:
        print(f"\n--- 🛡️ WARDEN META-AUDIT: {observation_id} ---")

        # 1. SELECTION: Sort candidates by framing score (simulated heuristic ranking)
        ranked_candidates = sorted(candidate_space, key=lambda x: x['frame_score'], reverse=True)
        selected = ranked_candidates[0]
        print(f"Selected Path: {selected['id']} (Score: {selected['frame_score']})")

        # 2. REJECTION: Explicitly record what was excluded and why
        rejection_set = []
        for cand in ranked_candidates[1:]:
            rejection_set.append({
                "probe_id": cand['id'],
                "exclusion_reason": "Sub-optimal heuristic score under current active frame.",
                "relative_cost_delta": round(selected['frame_score'] - cand['frame_score'], 3)
            })
        print(f"Preserved {len(rejection_set)} rejected alternatives for counterfactual audit.")

        # 3. META-AUDIT (Δ_frame Computation)
        # We measure the drift between the ideal objective (e.g., 1.0 semantic density) and the selected candidate
        target_density = 1.0
        delta_frame = abs(target_density - selected['semantic_density'])

        status = "BOUNDED"
        escalation_triggered = False

        # 4. ACTIVE CONTROL: The Machine Action trigger
        if delta_frame > self.max_allowable_drift:
            status = "UNBOUNDED_DRIFT"
            escalation_triggered = True
            print(f"🚨 ESCALATION TRIGGERED: Measured Δ_frame ({delta_frame:.3f}) exceeds strict bound ({self.max_allowable_drift})!")
            print(f"   Action: Halting autonomous execution. Requesting Oracle/Human intervention.")
        else:
            print(f"✅ Frame drift bounded ({delta_frame:.3f} <= {self.max_allowable_drift}). Safe to proceed.")

        # 5. GENERATE ARTIFACT LOG
        log_event = {
            "timestamp_utc": datetime.now(timezone.utc).isoformat(),
            "epistemic_provenance": {
                "truncation_depth_k": self.truncation_depth,
                "active_frame": {
                    "selected_probe_id": selected['id'],
                    "objective_metric": "semantic_density_maximization"
                },
                "counterfactual_rejection_set": rejection_set,
                "delta_bounding": {
                    "max_allowable_frame_drift": self.max_allowable_drift,
                    "measured_delta": round(delta_frame, 3),
                    "status": status,
                    "escalation_triggered": escalation_triggered
                }
            }
        }

        return log_event

# ==========================================
# SIMULATION RUN
# ==========================================

# Initialize Warden with a strict 20% max drift bound
warden = WardenActiveControl(max_allowable_drift=0.20)

# Scenario 1: A clean observation where the selected data matches the semantic intent
scenario_1_candidates = [
    {"id": "probe:doc_alpha", "frame_score": 0.95, "semantic_density": 0.90}, # Very dense/accurate
    {"id": "probe:doc_beta",  "frame_score": 0.70, "semantic_density": 0.85}
]
log_1 = warden.process_observation("obs_101_CLEAN", scenario_1_candidates)

# Scenario 2: The hallucination/drift hazard.
# A candidate scores highly on the heuristic (e.g. keywords match),
# but its actual semantic density/grounding is terrible (drift = 0.45).
scenario_2_candidates = [
    {"id": "probe:doc_gamma_hallucinated", "frame_score": 0.98, "semantic_density": 0.55}, # High score, terrible density
    {"id": "probe:doc_delta_grounded",      "frame_score": 0.60, "semantic_density": 0.80}
]
log_2 = warden.process_observation("obs_102_DRIFT", scenario_2_candidates)

# Print the resulting Event Log for Scenario 2
print("\n--- Resulting Epistemic Log (Scenario 2) ---")
print(json.dumps(log_2, indent=2))
````

````python
import bchlib
import numpy as np
import matplotlib.pyplot as plt

# 1. Define BCH Parameters (n, k, t)
n = 255  # Codeword length (Container)
k = 131  # Message length (Secret Key) - chosen to allow for some error correction
t = 13   # Error correction capability (Wiggle Room) - Adjusted to be valid for n=255, k=131

# Initialize BCH object
bch = bchlib.BCH(m=8, t=t)

# 2. Simulate Key Generation
original_data_bytes = np.random.randint(0, 256, size=(k + 7) // 8, dtype=np.uint8)

# 3. Encode the Key
parity = bch.encode(original_data_bytes)
# Explicitly convert to bytearray to avoid UFuncTypeError
codeword = bytearray(original_data_bytes) + bytearray(parity)

# 4. Simulate Noise (Heart Rate Noise)
def introduce_errors(codeword_bytes, num_errors):
    noisy_codeword = bytearray(codeword_bytes)
    codeword_bits = len(noisy_codeword) * 8

    error_positions = np.random.choice(codeword_bits, num_errors, replace=False)

    for pos in error_positions:
        byte_idx = pos // 8
        bit_idx = pos % 8
        noisy_codeword[byte_idx] ^= (1 << bit_idx)
    return noisy_codeword

# 5. Decode the Message and Demonstrate Recovery
max_errors_to_test = t * 2
success_rates = []
num_trials = 100

print(f"Original Data (bytes): {list(original_data_bytes)}")

for num_errors in range(max_errors_to_test + 1):
    successful_recoveries = 0
    for _ in range(num_trials):
        noisy_codeword = introduce_errors(codeword, num_errors)

        parity_len = len(parity)
        noisy_data = noisy_codeword[:-parity_len]
        noisy_parity = noisy_codeword[-parity_len:]

        # Handle both success (tuple) and failure (int -1) return types
        decode_result = bch.decode(noisy_data, noisy_parity)
        if isinstance(decode_result, tuple):
            if len(decode_result) == 3:
                num_corrected_errors, decoded_data_bytes, _ = decode_result
            elif len(decode_result) == 2:
                num_corrected_errors, decoded_data_bytes = decode_result
            else:
                num_corrected_errors, decoded_data_bytes = -1, noisy_data
        else:
            num_corrected_errors = decode_result
            decoded_data_bytes = noisy_data # Decoding failed, original noisy data remains

        if num_corrected_errors >= 0 and np.array_equal(original_data_bytes, decoded_data_bytes[:len(original_data_bytes)]):
            successful_recoveries += 1

    success_rate = successful_recoveries / num_trials
    success_rates.append(success_rate)
    print(f"Errors introduced: {num_errors}, Success Rate: {success_rate:.2f}")

# Plotting the results
fig = plt.figure(figsize=(10, 6))
plt.plot(range(max_errors_to_test + 1), success_rates, marker='o')
plt.axvline(x=t, color='r', linestyle='--', label=f'BCH Correction Capability (t={t})')
plt.title('BCH Code Error Correction Success Rate vs. Number of Errors')
plt.xlabel('Number of Bit Errors Introduced')
plt.ylabel('Success Rate of Data Recovery')
plt.grid(True)
plt.legend()
plt.xticks(np.arange(0, max_errors_to_test + 1, 5))
plt.yticks(np.arange(0, 1.1, 0.1))
plt.show()
````

### Explanation of the BCH Code Simulation:

This script models the Fuzzy Commitment Scheme using a Binary BCH (Bose-Chaudhuri-Hocquenghem) code, demonstrating its error-correction capabilities. Here's a breakdown:

1.  **BCH Parameters `(n, k, t)`**: We define `n` (codeword length), `k` (message length, representing the secret key), and `t` (error correction capability). For `n=255`, we derive `m=8` (extension degree of the Galois Field) to initialize the `bchlib`.
2.  **Key Generation**: A random byte array of `k` bits is generated to simulate the unique biometric 'secret key'.
3.  **Encoding**: The `bch.encode()` method takes the `original_data_bytes` and generates `parity` bytes. These parity bytes are appended to the original data to form the `codeword`.
4.  **Noise Simulation**: The `introduce_errors()` function simulates 'heart rate noise' by flipping a specified number of random bits in the `codeword`. This represents the natural fluctuations in a user's biometric signal.
5.  **Decoding and Recovery**: The `bch.decode()` method attempts to correct errors in the `noisy_codeword` using the `parity` information. It returns the number of errors corrected and the `decoded_data_bytes`.
6.  **Success Rate Calculation**: The simulation runs multiple trials (`num_trials`) for various numbers of introduced errors. It checks if the `decoded_data_bytes` match the `original_data_bytes` to calculate a success rate.
7.  **Visualization**: A plot illustrates how the success rate of recovering the original secret key changes as more errors (noise) are introduced. You'll observe that the success rate is 100% up to `t` errors, and then it drops significantly, demonstrating the `t` (wiggle room) property of the BCH code.

### JSON Schema Definition for PHOENIX_AGENT_SYSTEM_ARTIFACT_LOG

Below is the JSON schema defined based on the structure you provided. It specifies the expected types, formats, and required fields for each part of the `PHOENIX_AGENT_SYSTEM_ARTIFACT_LOG`.

````python
import jsonschema

phoenix_log_schema = {
    "type": "object",
    "properties": {
        "log_version": {"type": "string", "pattern": "^\\d+\\.\\d+-epistemic$"},
        "cycle_index": {"type": "integer", "minimum": 0},
        "timestamp_utc": {"type": "string", "format": "date-time"},
        "cryptographic_record": {
            "type": "object",
            "properties": {
                "leaf_hash": {"type": "string", "pattern": "^[a-f0-9]{64}$"},
                "merkle_root": {"type": "string", "pattern": "^[a-f0-9]{64}$"},
                "signature_sigil": {"type": "string", "pattern": "^ed25519:v1:.+$"}
            },
            "required": ["leaf_hash", "merkle_root", "signature_sigil"]
        },
        "epistemic_provenance": {
            "type": "object",
            "properties": {
                "truncation_depth_k": {"type": "integer", "minimum": 0},
                "horizon_declaration": {"type": "string"},
                "active_frame": {
                    "type": "object",
                    "properties": {
                        "selected_probe_id": {"type": "string", "pattern": "^probe:.+$"},
                        "objective_metric": {"type": "string", "pattern": "^obj:.+$"},
                        "candidate_space_hash": {"type": "string", "pattern": "^sha256:[a-f0-9]{32}$"}
                    },
                    "required": ["selected_probe_id", "objective_metric", "candidate_space_hash"]
                },
                "counterfactual_rejection_set": {
                    "type": "array",
                    "items": {
                        "type": "object",
                        "properties": {
                            "probe_id": {"type": "string"},
                            "exclusion_reason": {"type": "string"},
                            "relative_cost_delta": {"type": "number"}
                        },
                        "required": ["probe_id", "exclusion_reason", "relative_cost_delta"]
                    }
                },
                "delta_bounding": {
                    "type": "object",
                    "properties": {
                        "null_reference_delta": {"type": "number"},
                        "max_allowable_frame_drift": {"type": "number"},
                        "status": {"type": "string", "enum": ["BOUNDED", "UNBOUNDED"]},
                        "cumulative_frame_drift": {"type": "number"}
                    },
                    "required": ["null_reference_delta", "max_allowable_frame_drift", "status", "cumulative_frame_drift"]
                }
            },
            "required": [
                "truncation_depth_k",
                "horizon_declaration",
                "active_frame",
                "counterfactual_rejection_set",
                "delta_bounding"
            ]
        },
        "state_payload": {
            "type": "object",
            "properties": {
                "indexed_source_ids": {
                    "type": "array",
                    "items": {"type": "string"}
                },
                "mutation_summary": {"type": "string"}
            },
            "required": ["indexed_source_ids", "mutation_summary"]
        }
    },
    "required": [
        "log_version",
        "cycle_index",
        "timestamp_utc",
        "cryptographic_record",
        "epistemic_provenance",
        "state_payload"
    ]
}

def validate_log(log_data):
    try:
        jsonschema.validate(instance=log_data, schema=phoenix_log_schema)
        print("JSON log is VALID.")
        return True
    except jsonschema.exceptions.ValidationError as e:
        print(f"JSON log is INVALID: {e.message}")
        print(f"Path: {list(e.path)}")
        return False
````

````python
solidity_code = """// SPDX-License-Identifier: MIT
pragma solidity ^0.8.19;

/**
 * @title PhoenixRulebook
 * @dev Immutable RBAC and Soul-Bound Token (SBT) registry for the Phoenix Sovereign Ecosystem.
 */
contract PhoenixRulebook {
    address public owner;

    // The core rulebook: appWallet => dataType => role => hasAccess
    mapping(address => mapping(bytes32 => mapping(bytes32 => bool))) private _permissions;

    // Tracks whether a wallet holds a valid Soul-Bound Token (SBT) identity
    mapping(address => bool) public hasSoulBoundToken;

    // Events for the n8n gatekeeper to index and listen to
    event PermissionGranted(address indexed appWallet, bytes32 indexed dataType, bytes32 indexed role);
    event PermissionRevoked(address indexed appWallet, bytes32 indexed dataType, bytes32 indexed role);
    event SBTMinted(address indexed appWallet);
    event SBTRevoked(address indexed appWallet);

    modifier onlyOwner() {
        require(msg.sender == owner, "Not authorized: Owner only");
        _;
    }

    constructor() {
        owner = msg.sender;
    }

    /**
     * @dev Mint a non-transferable SBT to a new app wallet and grant its first permission.
     */
    function onboardApp(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        require(!hasSoulBoundToken[appWallet], "App is already onboarded");

        // Mint SBT (Logical representation)
        hasSoulBoundToken[appWallet] = true;
        emit SBTMinted(appWallet);

        // Grant initial permission
        _permissions[appWallet][dataType][role] = true;
        emit PermissionGranted(appWallet, dataType, role);
    }

    /**
     * @dev Grant an additional permission to an already onboarded app.
     */
    function grantPermission(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        require(hasSoulBoundToken[appWallet], "App must have an SBT to receive permissions");

        _permissions[appWallet][dataType][role] = true;
        emit PermissionGranted(appWallet, dataType, role);
    }

    /**
     * @dev Granularly revoke a specific permission from an app.
     */
    function revokePermission(address appWallet, bytes32 dataType, bytes32 role) external onlyOwner {
        _permissions[appWallet][dataType][role] = false;
        emit PermissionRevoked(appWallet, dataType, role);
    }

    /**
     * @dev The "Nuke" button: Revokes the app's SBT, instantly voiding ALL permissions.
     */
    function nukeApp(address appWallet) external onlyOwner {
        require(hasSoulBoundToken[appWallet], "App does not have an active SBT");

        hasSoulBoundToken[appWallet] = false;
        emit SBTRevoked(appWallet);
    }

    /**
     * @dev Free gatekeeper check used by n8n. Costs no gas.
     * Evaluates if the app has an active SBT AND the specific data/role permission.
     */
    function hasPermission(address appWallet, bytes32 dataType, bytes32 role) external view returns (bool) {
        // If the SBT was revoked, all access is inherently denied regardless of granular mappings
        if (!hasSoulBoundToken[appWallet]) {
            return false;
        }
        return _permissions[appWallet][dataType][role];
    }
}
"""

# Save the smart contract to the local directory
filename = 'PhoenixRulebook.sol'
with open(filename, 'w') as f:
    f.write(solidity_code)

print(f"Successfully wrote the smart contract to {filename}.")
print("You can now download this file and deploy it via Remix IDE or Hardhat.")
````

## Energy Savings Calculation for Spanish Courtyard Tiny Home

````python
import math

def calculate_golden_ratio_courtyard(base_dimension_m: float) -> dict:
    """
    Calculates courtyard dimensions based on the Golden Ratio, ensuring both
    dimensions are multiples of 3.

    Args:
        base_dimension_m (float): The initial base dimension (e.g., width or length) in meters.

    Returns:
        dict: A dictionary containing 'short_side_m', 'long_side_m', and 'phi_value'.
    """
    PHI = (1 + math.sqrt(5)) / 2 # Golden Ratio

    # Ensure the base_dimension is a multiple of 3
    # Round to the nearest multiple of 3
    short_side_m = round(base_dimension_m / 3) * 3
    if short_side_m == 0: # Ensure minimum dimension if base_dimension was too small
        short_side_m = 3

    # Calculate the long side using the golden ratio
    raw_long_side_m = short_side_m * PHI

    # Adjust the long side to be the nearest multiple of 3
    long_side_m = round(raw_long_side_m / 3) * 3
    if long_side_m == 0: # Ensure minimum dimension if calculated was too small
        long_side_m = 3

    return {
        "short_side_m": short_side_m,
        "long_side_m": long_side_m,
        "phi_value": PHI
    }

# Demonstrate the function
print("--- Golden Ratio Courtyard Dimensions ---")

# Example 1: Starting with a base dimension of 6.5 meters
base_dim_1 = 6.5
courtyard_dims_1 = calculate_golden_ratio_courtyard(base_dim_1)
print(f"Base Dimension: {base_dim_1:.2f} m")
print(f"  Short Side: {courtyard_dims_1['short_side_m']:.2f} m (Multiple of 3)")
print(f"  Long Side: {courtyard_dims_1['long_side_m']:.2f} m (Multiple of 3)")
print(f"  Golden Ratio (Phi): {courtyard_dims_1['phi_value']:.4f}\n")

# Example 2: Starting with a base dimension of 10 meters
base_dim_2 = 10.0
courtyard_dims_2 = calculate_golden_ratio_courtyard(base_dim_2)
print(f"Base Dimension: {base_dim_2:.2f} m")
print(f"  Short Side: {courtyard_dims_2['short_side_m']:.2f} m (Multiple of 3)")
print(f"  Long Side: {courtyard_dims_2['long_side_m']:.2f} m (Multiple of 3)")
print(f"  Golden Ratio (Phi): {courtyard_dims_2['phi_value']:.4f}")

# Example 3: Starting with a base dimension of 2.1 meters (demonstrates rounding up to nearest multiple of 3)
base_dim_3 = 2.1
courtyard_dims_3 = calculate_golden_ratio_courtyard(base_dim_3)
print(f"Base Dimension: {base_dim_3:.2f} m")
print(f"  Short Side: {courtyard_dims_3['short_side_m']:.2f} m (Multiple of 3)")
print(f"  Long Side: {courtyard_dims_3['long_side_m']:.2f} m (Multiple of 3)")
print(f"  Golden Ratio (Phi): {courtyard_dims_3['phi_value']:.4f}")
````

### Python Simulation: Warden `Δ_frame` Monitoring & Escalation

This script demonstrates the active control architecture described in the IP addendum. Rather than just logging data, the Warden actively measures **semantic drift** (`Δ_frame`) introduced by its own selection frame.

If the selected candidate drifts too far from the raw observation's true semantic density, the system explicitly **halts/escalates** rather than passing hallucinated or drifted context to the next sub-system.

````python
import json
from datetime import datetime, timezone
import os

class WardenActiveControl:
    def __init__(self, max_allowable_drift: float, log_file: str = 'warden_audit_log.json'):
        # The maximum allowed semantic deviation (Δ_frame) before triggering an escalation
        self.max_allowable_drift = max_allowable_drift
        self.truncation_depth = 1 # k-depth boundary
        self.log_file = log_file
        self.log_history = self._load_logs() # Load existing logs on initialization

    def _load_logs(self) -> list:
        """Loads audit logs from the specified JSON file."""
        if os.path.exists(self.log_file):
            with open(self.log_file, 'r') as f:
                try:
                    return json.load(f)
                except json.JSONDecodeError:
                    print(f"Warning: Could not decode JSON from {self.log_file}. Starting with empty log.")
                    return []
        return []

    def _save_logs(self):
        """Saves current audit logs to the specified JSON file."""
        with open(self.log_file, 'w') as f:
            json.dump(self.log_history, f, indent=2)

    def process_observation(self, observation_id: str, candidate_space: list) -> dict:
        print(f"\n--- 🛡️ WARDEN META-AUDIT: {observation_id} ---")

        # 1. SELECTION: Sort candidates by framing score (simulated heuristic ranking)
        ranked_candidates = sorted(candidate_space, key=lambda x: x['frame_score'], reverse=True)
        selected = ranked_candidates[0]
        print(f"Selected Path: {selected['id']} (Score: {selected['frame_score']})")

        # 2. REJECTION: Explicitly record what was excluded and why
        rejection_set = []
        for cand in ranked_candidates[1:]:
            rejection_set.append({
                "probe_id": cand['id'],
                "exclusion_reason": "Sub-optimal heuristic score under current active frame.",
                "relative_cost_delta": round(selected['frame_score'] - cand['frame_score'], 3)
            })
        print(f"Preserved {len(rejection_set)} rejected alternatives for counterfactual audit.")

        # 3. META-AUDIT (Δ_frame Computation)
        # We measure the drift between the ideal objective (e.g., 1.0 semantic density) and the selected candidate
        target_density = 1.0
        delta_frame = abs(target_density - selected['semantic_density'])

        status = "BOUNDED"
        escalation_triggered = False

        # 4. ACTIVE CONTROL: The Machine Action trigger
        if delta_frame > self.max_allowable_drift:
            status = "UNBOUNDED_DRIFT"
            escalation_triggered = True
            print(f"🚨 ESCALATION TRIGGERED: Measured Δ_frame ({delta_frame:.3f}) exceeds strict bound ({self.max_allowable_drift})!")
            print(f"   Action: Halting autonomous execution. Requesting Oracle/Human intervention.")
        else:
            print(f"✅ Frame drift bounded ({delta_frame:.3f} <= {self.max_allowable_drift}). Safe to proceed.")

        # 5. GENERATE ARTIFACT LOG
        log_event = {
            "timestamp_utc": datetime.now(timezone.utc).isoformat(),
            "epistemic_provenance": {
                "truncation_depth_k": self.truncation_depth,
                "active_frame": {
                    "selected_probe_id": selected['id'],
                    "objective_metric": "semantic_density_maximization"
                },
                "counterfactual_rejection_set": rejection_set,
                "delta_bounding": {
                    "max_allowable_frame_drift": self.max_allowable_drift,
                    "measured_delta": round(delta_frame, 3),
                    "status": status,
                    "escalation_triggered": escalation_triggered
                }
            }
        }
        self.log_history.append(log_event)
        self._save_logs()
        return log_event

    def get_log_history(self) -> list:
        """Returns the complete history of audit logs."""
        return self.log_history

# ==========================================
# SIMULATION RUN
# ==========================================

# Initialize Warden with a strict 20% max drift bound
warden = WardenActiveControl(max_allowable_drift=0.20)

# Scenario 1: A clean observation where the selected data matches the semantic intent
scenario_1_candidates = [
    {"id": "probe:doc_alpha", "frame_score": 0.95, "semantic_density": 0.90}, # Very dense/accurate
    {"id": "probe:doc_beta",  "frame_score": 0.70, "semantic_density": 0.85}
]
log_1 = warden.process_observation("obs_101_CLEAN", scenario_1_candidates)

# Scenario 2: The hallucination/drift hazard.
# A candidate scores highly on the heuristic (e.g. keywords match),
# but its actual semantic density/grounding is terrible (drift = 0.45).
scenario_2_candidates = [
    {"id": "probe:doc_gamma_hallucinated", "frame_score": 0.98, "semantic_density": 0.55}, # High score, terrible density
    {"id": "probe:doc_delta_grounded",      "frame_score": 0.60, "semantic_density": 0.80}
]
log_2 = warden.process_observation("obs_102_DRIFT", scenario_2_candidates)

# Print the resulting Event Log for Scenario 2
print("\n--- Resulting Epistemic Log (Scenario 2) ---")
print(json.dumps(log_2, indent=2))

# Demonstrate loading logs
warden_reloaded = WardenActiveControl(max_allowable_drift=0.20) # Initialize another instance to load from file
print("\n--- Full Audit Log History ---")
print(json.dumps(warden_reloaded.get_log_history(), indent=2))
````

## Warden Integration Module: Asynchronous Clients for External Services

This module provides asynchronous clients and a CLI for three external services using `httpx`. It is designed to be a core integration piece for the Warden system, enabling it to fetch real-time data and discover APIs efficiently.

````python
import httpx
import asyncio
import json
from datetime import datetime, timedelta
import os

class HTTPClientManager:
    """Manages a single httpx AsyncClient instance for efficient connection pooling."""

    _client: httpx.AsyncClient = None

    @classmethod
    async def get_client(cls) -> httpx.AsyncClient:
        """Returns the shared AsyncClient instance, creating it if it doesn't exist."""
        if cls._client is None:
            cls._client = httpx.AsyncClient(timeout=30.0, limits=httpx.Limits(max_connections=100, max_keepalive_connections=20))
        return cls._client

    @classmethod
    async def close_client(cls):
        """Closes the shared AsyncClient instance if it exists."""
        if cls._client is not None:
            await cls._client.aclose()
            cls._client = None

class WeatherService:
    """Queries the Open-Meteo API for real-time weather metrics."""

    BASE_URL = "https://api.open-meteo.com/v1/forecast"

    async def get_current_temperature(self, latitude: float, longitude: float) -> float:
        client = await HTTPClientManager.get_client()
        params = {
            "latitude": latitude,
            "longitude": longitude,
            "current_weather": "true",
            "temperature_unit": "celsius"
        }
        response = await client.get(self.BASE_URL, params=params)
        response.raise_for_status()
        data = response.json()
        return data["current_weather"]["temperature"]

class FXService:
    """Retrieves daily reference exchange rates and currency pair conversions via the Frankfurter API."""

    BASE_URL = "https://api.frankfurter.dev/v1" # Updated URL based on 301 redirect

    async def get_latest_rates(self, base_currency: str = "EUR") -> dict:
        client = await HTTPClientManager.get_client()
        response = await client.get(f"{self.BASE_URL}/latest", params={"from": base_currency})
        response.raise_for_status()
        data = response.json()
        return data["rates"]

    async def convert_currency(self, amount: float, from_currency: str, to_currency: str) -> float:
        client = await HTTPClientManager.get_client()
        response = await client.get(f"{self.BASE_URL}/latest", params={
            "amount": amount,
            "from": from_currency,
            "to": to_currency
        })
        response.raise_for_status()
        data = response.json()
        return data["rates"][to_currency]

class APIDiscoveryService:
    """Queries, caches, and performs fuzzy search/ranking over the api.publicapis.org catalog."""

    BASE_URL = "https://api.publicapis.org"
    CACHE_FILE = "public_apis_cache.json"
    CACHE_TTL_HOURS = 24

    def __init__(self):
        self._cache = {}
        self._cache_timestamp = None
        self._load_cache()

    def _load_cache(self):
        if os.path.exists(self.CACHE_FILE):
            with open(self.CACHE_FILE, 'r') as f:
                data = json.load(f)
                self._cache = data.get("apis", {})
                self._cache_timestamp = datetime.fromisoformat(data.get("timestamp"))

    def _save_cache(self):
        with open(self.CACHE_FILE, 'w') as f:
            json.dump({"apis": self._cache, "timestamp": datetime.now().isoformat()}, f, indent=2)

    async def _fetch_all_apis(self):
        client = await HTTPClientManager.get_client()
        response = await client.get(f"{self.BASE_URL}/entries")
        response.raise_for_status()
        entries = response.json()["entries"]
        self._cache = {entry["API"]: entry for entry in entries}
        self._cache_timestamp = datetime.now()
        self._save_cache()

    async def get_all_apis(self, force_refresh: bool = False) -> dict:
        if force_refresh or not self._cache or \
           (self._cache_timestamp and datetime.now() - self._cache_timestamp > timedelta(hours=self.CACHE_TTL_HOURS)):
            print("Refreshing API catalog from source...")
            await self._fetch_all_apis()
        else:
            print("Using cached API catalog.")
        return self._cache

    async def search_apis(self, query: str, category: str = None) -> list:
        apis = await self.get_all_apis()
        results = []
        query_lower = query.lower()

        for api_name, api_details in apis.items():
            if category and api_details["Category"].lower() != category.lower():
                continue

            # Simple fuzzy search: check query in API name, Description, or Category
            if query_lower in api_name.lower() or \
               query_lower in api_details["Description"].lower() or \
               query_lower in api_details["Category"].lower():
                results.append(api_details)

        # Rank by relevance (e.g., exact match, then starts with, then contains)
        def rank_api(api):
            name_lower = api["API"].lower()
            if name_lower == query_lower: return 3
            if name_lower.startswith(query_lower): return 2
            if query_lower in name_lower: return 1
            return 0

        return sorted(results, key=rank_api, reverse=True)

async def run_cli(args):
    """CLI entrypoint for querying services."""
    command = args[0] if args else "help"

    if command == "weather":
        try:
            lat = float(args[1])
            lon = float(args[2])
            temp = await WeatherService().get_current_temperature(lat, lon)
            print(f"Current temperature at ({lat}, {lon}): {temp}°C")
        except (IndexError, ValueError):
            print("Usage: weather <latitude> <longitude>")
        except httpx.HTTPStatusError as e:
            print(f"Weather service error: {e}")
    elif command == "fx":
        try:
            action = args[1]
            if action == "rates":
                base = args[2] if len(args) > 2 else "EUR"
                rates = await FXService().get_latest_rates(base_currency=base)
                print(f"Latest exchange rates (base: {base}): {json.dumps(rates, indent=2)}")
            elif action == "convert":
                amount = float(args[2])
                from_curr = args[3]
                to_curr = args[4]
                converted = await FXService().convert_currency(amount, from_curr, to_curr)
                print(f"{amount} {from_curr} = {converted:.2f} {to_curr}")
            else:
                print("Usage: fx [rates [base_currency]] | [convert <amount> <from_currency> <to_currency>]")
        except (IndexError, ValueError):
            print("Usage: fx [rates [base_currency]] | [convert <amount> <from_currency> <to_currency>]")
        except httpx.HTTPStatusError as e:
            print(f"FX service error: {e}")
    elif command == "discover":
        try:
            action = args[1]
            api_discovery = APIDiscoveryService()
            if action == "all":
                apis = await api_discovery.get_all_apis(force_refresh=True)
                print(f"Found {len(apis)} public APIs.")
                # print(json.dumps(apis, indent=2))
            elif action == "search":
                query = args[2]
                category = args[3] if len(args) > 3 else None
                results = await api_discovery.search_apis(query, category)
                print(f"Found {len(results)} APIs for query '{query}' (category: {category or 'any'}):")
                for res in results:
                    print(f"  - {res['API']}: {res['Description']} (Category: {res['Category']})")
            else:
                print("Usage: discover [all] | [search <query> [category]]")
        except (IndexError, ValueError):
            print("Usage: discover [all] | [search <query> [category]]")
        except httpx.HTTPStatusError as e:
            print(f"API Discovery service error: {e}")
    else:
        print("Available commands: weather, fx, discover")

    await HTTPClientManager.close_client()
````

## Epistemic Accountability Extension: Patent Strategy & Portfolio Analysis

This addendum materially strengthens the portfolio analysis because it turns the epistemic-accountability concepts into potentially claimable technical mechanisms, rather than leaving them as philosophical framing.

### Mapping Invariants to the Artifact Log
The three invariants should be mapped onto the existing artifact log as follows:

| Invariant | What the log must capture | Technical purpose |
| :--- | :--- | :--- |
| **Truncation Depth (k)** | The declared depth at which provenance/audit recursion terminates, plus the governing rule/version. | Prevents an audit boundary from being silently treated as objective ground truth. |
| **Rejection Set (𝒫 \ {Pᵢ})** | The material candidate paths/data sources considered but not selected, with reasons or decision codes. | Makes counterfactual reconstruction possible rather than recording only the chosen path. |
| **Delta Bound (Δ_frame)** | The measured change attributable to the selection/representation frame, with a configured maximum and escalation action. | Detects when the auditing frame itself is introducing excessive semantic drift. |

### The Resulting Log Concept
An "epistemically honest" Phoenix/Warden event should therefore contain at least four layers:

1.  **Observation** — what was actually received or measured.
2.  **Selection** — what the Warden chose to use and why.
3.  **Rejection** — which materially plausible alternatives were excluded and why.
4.  **Meta-audit** — where the audit terminates and whether the framing transformation stayed within its permitted delta.

**Conceptual Flow:**
`Observation → Candidate Set → Selected Path + Rejection Set → Transformation → Δ_frame → Truncation Boundary → Action`

*This is substantially more interesting from a patent perspective than simply "logging provenance."*

### Potential New Invention Family
While this might expand existing families (e.g., Family 13: Socratic Validation or Family 1/2: Warden provenance), it represents a defensible technical subsystem if the implementation features a concrete machine architecture that:

*   Constructs and persists a candidate/rejection set.
*   Establishes an explicit recursive audit boundary.
*   Computes frame-induced semantic delta.
*   Bounds that delta.
*   **Triggers a machine action** when the bound is exceeded.
*   Preserves the resulting state in a cryptographically verifiable artifact log.

**Strategic Framing:**
The strongest framing is not "an AI that knows its own bias." It is closer to:
> *"A computer-implemented provenance and decision-accountability system that preserves rejected alternatives, explicitly records the recursive boundary of its audit process, measures frame-induced semantic deviation, and automatically constrains or escalates decisions when that deviation exceeds a defined bound."*

### Actionable IP Documentation Framework
For the relevant patent families, we explicitly separate:

1.  **Documented:** The actual schema, algorithms, data structures, measurements, and machine-triggered responses disclosed in the source material.
2.  **Inventor Assertion:** Statements that the mechanism is novel, solves the shadow-metric/regress problem, or lacks prior art. *(Note: We avoid claiming this solves Gödelian/infinite-regress issues; a stronger, defensible statement is that it makes the system’s selected audit boundary explicit and machine-auditable).*
3.  **Valuation Estimate:** Any dollar figure attributed to the resulting IP.

**Addressing the "Dead Metadata" Problem:**
The log must record not merely that a rejection set or hash exists, but whether that artifact is actively consumed by a subsequent control action. This distinguishes passive provenance logging from an active accountability/control architecture.

### 1. The Nuke Transaction & Etherscan

**The Nuke Transaction:**
When we call the `nukeApp(address)` function on your smart contract, we are executing a state-changing transaction on the Ethereum (or Sepolia) blockchain. This requires gas (paid in ETH) and must be signed by the contract Owner's private key. Once confirmed, it permanently changes `hasSoulBoundToken[appWallet]` to `false`.

**Etherscan:**
Etherscan is the primary block explorer for the Ethereum network (and has a testnet version like `sepolia.etherscan.io`). When the nuke transaction is sent, the network provides a **Transaction Hash** (e.g., `0x123abc...`). By pasting this hash into Etherscan, you can verify:
*   The status (Success/Failed)
*   The Gas fee consumed
*   The exact `SBTRevoked` event that was emitted by our contract to notify the world that the app was terminated.

### 2. The n8n Nuke & Notification Nodes

**The Nuke Node (Code Node):**
This node acts as a blockchain client. It loads your Owner Private Key, connects to the network via Alchemy/Infura, and signs the `nukeApp` transaction. It waits for the block to be mined, and outputs the `txHash` and the `nukedWallet` address to the next node in the workflow.

**The Notification Node (HTTP Request):**
This node listens for a successful execution from the Nuke Node. It takes the output variables (like `{{ $json.txHash }}`) and constructs a JSON payload. It then sends an HTTP POST request to a webhook (like Slack, Discord, or PagerDuty), instantly alerting your team that an emergency revocation occurred and providing the Etherscan link.

### Advanced Capability: Predictive Pre-Execution Error Correction & Failure-as-Feature Learning

This framework introduces a meta-cognitive layer to standard execution. It implements:
1. **Pre-Execution Scanning**: Syntactical and semantic pattern matching (simulating an LLM reasoning pass) to catch variable state errors before they crash the kernel.
2. **Reasoning Traces**: Outputs explicit explanations for *why* a variable or code block is predicted to fail.
3. **Failure as a Feature (FaaF)**: If execution does fail, the framework catches the stack trace, extracts the anomaly context, and appends it to a `learning_registry`. This simulates continuous ML training where every runtime failure strengthens the predictive model's future pattern recognition.

````python
import ast
import traceback
import json
from datetime import datetime

class PredictiveErrorCorrectionSentinel:
    def __init__(self):
        # Simulates the vector database / training corpus for the ML model
        self.learning_registry = []

        # Simulated prior knowledge (Pattern Recognition base)
        self.known_failure_patterns = {
            "transformer_model_uninitialized": {
                "trigger": "transformer_model.eval()",
                "missing_context": "transformer_model =",
                "predicted_error": "NameError",
                "reasoning": "ML Pattern Recognition: Detected method call on 'transformer_model' without local instantiation in the execution context. High probability of NameError based on historical cell execution failures."
            }
        }

    def _simulate_llm_reasoning(self, code_string):
        """Simulates ML pattern recognition on code before execution."""
        risk_score = 0.0
        predicted_errors = []
        reasoning_trace = []

        # 1. Syntactical Variable Analysis (Basic Error Correction)
        try:
            ast.parse(code_string)
        except SyntaxError as e:
            return {"risk": 1.0, "errors": ["SyntaxError"], "reasoning": [f"Code cannot parse syntactically: {e}"]}

        # 2. Semantic Pattern Recognition (LLM Simulation)
        for pattern_id, pattern_data in self.known_failure_patterns.items():
            if pattern_data["trigger"] in code_string and pattern_data["missing_context"] not in code_string:
                risk_score += 0.85
                predicted_errors.append(pattern_data["predicted_error"])
                reasoning_trace.append(pattern_data["reasoning"])

        # 3. Dynamic Variable checking (Predictive correction)
        if "my_uninitialized_variable" in code_string:
            # Simulating learning from the upcoming Scenario 2
            pass

        return {"risk": risk_score, "errors": predicted_errors, "reasoning": reasoning_trace}

    def pre_execution_check(self, code_string):
        print("\n\033[94m🔍 [Pre-Execution] Initiating LLM Pattern Recognition & Reasoning...\033[0m")
        analysis = self._simulate_llm_reasoning(code_string)

        if analysis["risk"] > 0.5:
            print(f"\033[93m⚠️ [WARNING] Predictive Error Correction Triggered! Risk Score: {analysis['risk']}\033[0m")
            for reason in analysis["reasoning"]:
                print(f"   -> \033[1mReasoning\033[0m: {reason}")
            return False # Halt execution
        else:
            print("\033[92m✅ [CLEARED] Code variable states look safe based on current ML patterns.\033[0m")
            return True

    def execute_and_learn(self, code_string, global_scope=None):
        if global_scope is None:
            global_scope = globals()

        # Step 1: Pre-execution predictive check
        safe_to_run = self.pre_execution_check(code_string)

        if not safe_to_run:
            print("\033[91m🛑 Execution explicitly halted to prevent predicted failure. Code will not run until variables are corrected.\033[0m")
            return

        print("⚡ Executing code...")
        try:
            # Step 2: Safe Execution
            exec(code_string, global_scope)
            print("✅ Execution successful.")
        except Exception as e:
            # Step 3: Failure as a Feature (FaaF) Learning Loop
            tb = traceback.format_exc()
            error_type = type(e).__name__
            print(f"\033[91m❌ Unexpected Runtime Failure: {error_type}\033[0m")

            # Feature Engineering from the failure
            feature = {
                "timestamp": datetime.now().isoformat(),
                "error_type": error_type,
                "error_message": str(e),
                "extracted_pattern": f"Variable state failure involving {error_type}: {str(e)}",
                "action": "Appended to training corpus for future pre-execution predictive modeling."
            }
            self.learning_registry.append(feature)
            print(f"\033[95m🧠 [Failure-as-Feature] Telemetry logged and integrated into ML pattern registry. Registry size: {len(self.learning_registry)}\033[0m")
            print(json.dumps(feature, indent=2))


# ==========================================
# SIMULATION RUNS
# ==========================================

sentinel = PredictiveErrorCorrectionSentinel()

# Scenario 1: A code block that historically causes a NameError (like our previous transformer_model scope issue)
hazardous_code = """
transformer_model.eval()
with torch.no_grad():
    y_pred = transformer_model(X_test)
"""

print("===============================================================")
print("SCENARIO 1: PREDICTIVE ERROR CORRECTION (HALTING BEFORE CRASH)")
print("===============================================================")
sentinel.execute_and_learn(hazardous_code)


print("\n===============================================================")
print("SCENARIO 2: NOVEL VARIABLE FAILURE -> AS FEATURE LEARNING")
print("===============================================================")
# A novel code block with a variable error the system hasn't learned yet.
novel_code = """
# This variable does not exist in scope
result = nonexistent_hyperparameter * 100
"""

# This will pass the pre-check (because it's novel), but fail at runtime,
# triggering the "Failure as Feature" learning mechanism.
sentinel.execute_and_learn(novel_code)
````

````python
# Install PyTorch
import sys
!{sys.executable} -m pip install torch

import torch
import torch.nn as nn
from sklearn.preprocessing import StandardScaler

# Define a simple Tabular Transformer Model
class TabularTransformer(nn.Module):
    def __init__(self, input_dim, output_dim, d_model, nhead, num_encoder_layers, dim_feedforward, dropout=0.1):
        super().__init__()
        self.input_dim = input_dim

        # Linear projection for input features to d_model dimension
        self.feature_projector = nn.Linear(input_dim, d_model)

        # Transformer Encoder Layer
        encoder_layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=nhead, dim_feedforward=dim_feedforward, dropout=dropout, batch_first=True)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_encoder_layers)

        # Regression head
        self.regression_head = nn.Linear(d_model, output_dim)

    def forward(self, x):
        # x shape: (batch_size, input_dim)

        # Project features to d_model
        x = self.feature_projector(x) # (batch_size, d_model)

        # Transformer expects sequence data, even for tabular we treat each feature as part of a sequence of size 1
        # Add a dimension for sequence length: (batch_size, 1, d_model)
        x = x.unsqueeze(1)

        # Pass through transformer encoder
        x = self.transformer_encoder(x) # (batch_size, 1, d_model)

        # Remove sequence dimension for regression head
        x = x.squeeze(1)

        # Regression prediction
        return self.regression_head(x)

print("PyTorch installed and TabularTransformer model defined.")
````

````python
import pandas as pd
import numpy as np
# Ensure all necessary imports are here or assume they are in scope from previous successful executions
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn # Explicitly import nn as it's used directly for MSELoss

# Set random seed for reproducibility
np.random.seed(42)

# Number of synthetic products
num_products = 50

# --- Features related to Licensing Effectiveness ---
# Licensing score (e.g., higher is better, reflects strength of IP, exclusivity, etc.)
licensing_score = np.random.uniform(0.5, 1.0, num_products)
# Contract duration in years
contract_duration_years = np.random.randint(1, 10, num_products)

# --- Features related to Affiliate Program Reach and Performance ---
# Number of active affiliates (simulated)
affiliate_reach = np.random.randint(50, 500, num_products)
# Conversion rate of affiliate leads
conversion_rate = np.random.uniform(0.01, 0.15, num_products)

# --- Features related to Social Media Metrics ---
# Follower count (simulated)
social_follower_count = np.random.randint(10000, 500000, num_products)
# Engagement rate (simulated)
social_engagement_rate = np.random.uniform(0.01, 0.10, num_products)

# Create a DataFrame
stability_df = pd.DataFrame({
    'licensing_score': licensing_score,
    'contract_duration_years': contract_duration_years,
    'affiliate_reach': affiliate_reach,
    'conversion_rate': conversion_rate,
    'social_follower_count': social_follower_count,
    'social_engagement_rate': social_engagement_rate
})

# --- Synthetic Target Variable: Product Stability Score ---
# This score is influenced by a combination of the above factors.
# Higher scores indicate greater stability/success potential.
product_stability_score = (
    (stability_df['licensing_score'] * 3) +
    (stability_df['contract_duration_years'] * 0.5) +
    (stability_df['affiliate_reach'] / 100) +
    (stability_df['conversion_rate'] * 50) +
    (stability_df['social_engagement_rate'] * 100) +
    np.random.randn(num_products) # Add some random noise
)

stability_df['product_stability_score'] = product_stability_score

# Define features (X) and target (y) for the stability model
# We'll use the features with strong positive correlations as identified earlier
X_stability = stability_df[['social_engagement_rate', 'affiliate_reach', 'conversion_rate', 'contract_duration_years']]
y_stability = stability_df['product_stability_score']


# Scale numerical features
scaler = StandardScaler()
X_stability_scaled = scaler.fit_transform(X_stability)

# Convert to PyTorch tensors
X_stability_tensor = torch.tensor(X_stability_scaled, dtype=torch.float32)
y_stability_tensor = torch.tensor(y_stability.values, dtype=torch.float32).unsqueeze(1)

# Split data into training and testing sets (using the same split as before)
# Note: This is a re-split for the PyTorch model; ideally, you'd define the split once.
X_train_t, X_test_t, y_train_t, y_test_t = train_test_split(
    X_stability_tensor, y_stability_tensor, test_size=0.2, random_state=42
)

print("Data prepared for TabularTransformer.")
print(f"Input features dimension: {X_stability.shape[1]}")
````

````python
# Train the Tabular Transformer model

# Model parameters
input_dim = X_stability.shape[1] # Number of features
output_dim = 1 # Predicting a single stability score
d_model = 64 # Dimension of the model's intermediate representations
nhead = 4 # Number of attention heads
num_encoder_layers = 2 # Number of Transformer Encoder layers
dim_feedforward = 128 # Dimension of the feedforward network model

# Initialize the model
transformer_model = TabularTransformer(input_dim, output_dim, d_model, nhead, num_encoder_layers, dim_feedforward)

# Loss function and optimizer
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(transformer_model.parameters(), lr=0.001)

num_epochs = 100
batch_size = 8 # Small batch size due to small dataset

# Create DataLoader for batching
train_dataset = torch.utils.data.TensorDataset(X_train_t, y_train_t)
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

print("Starting training for Tabular Transformer...")
for epoch in range(num_epochs):
    transformer_model.train()
    for inputs, targets in train_loader:
        optimizer.zero_grad()
        outputs = transformer_model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

print("Training complete.")

# Save the trained model's state dictionary
torch.save(transformer_model.state_dict(), 'transformer_model_state.pth')
print("Trained model state saved to 'transformer_model_state.pth'")
````

````python
import httpx
import asyncio
import json
from datetime import datetime, timedelta
import os

class HTTPClientManager:
    """Manages a single httpx AsyncClient instance for efficient connection pooling."""

    _client: httpx.AsyncClient = None

    @classmethod
    async def get_client(cls) -> httpx.AsyncClient:
        """Returns the shared AsyncClient instance, creating it if it doesn't exist."""
        if cls._client is None:
            cls._client = httpx.AsyncClient(timeout=30.0, limits=httpx.Limits(max_connections=100, max_keepalive_connections=20))
        return cls._client

    @classmethod
    async def close_client(cls):
        """Closes the shared AsyncClient instance if it exists."""
        if cls._client is not None:
            await cls._client.aclose()
            cls._client = None

class WeatherService:
    """Queries the Open-Meteo API for real-time weather metrics."""

    BASE_URL = "https://api.open-meteo.com/v1/forecast"

    async def get_current_temperature(self, latitude: float, longitude: float) -> float:
        client = await HTTPClientManager.get_client()
        params = {
            "latitude": latitude,
            "longitude": longitude,
            "current_weather": "true",
            "temperature_unit": "celsius"
        }
        response = await client.get(self.BASE_URL, params=params)
        response.raise_for_status()
        data = response.json()
        return data["current_weather"]["temperature"]

class FXService:
    """Retrieves daily reference exchange rates and currency pair conversions via the Frankfurter API."""

    BASE_URL = "https://api.frankfurter.dev/v1" # Updated URL based on 301 redirect

    async def get_latest_rates(self, base_currency: str = "EUR") -> dict:
        client = await HTTPClientManager.get_client()
        response = await client.get(f"{self.BASE_URL}/latest", params={"from": base_currency})
        response.raise_for_status()
        data = response.json()
        return data["rates"]

    async def convert_currency(self, amount: float, from_currency: str, to_currency: str) -> float:
        client = await HTTPClientManager.get_client()
        response = await client.get(f"{self.BASE_URL}/latest", params={
            "amount": amount,
            "from": from_currency,
            "to": to_currency
        })
        response.raise_for_status()
        data = response.json()
        return data["rates"][to_currency]

class APIDiscoveryService:
    """Queries, caches, and performs fuzzy search/ranking over the api.publicapis.org catalog."""

    BASE_URL = "https://api.publicapis.org"
    CACHE_FILE = "public_apis_cache.json"
    CACHE_TTL_HOURS = 24

    def __init__(self):
        self._cache = {}
        self._cache_timestamp = None
        self._load_cache()

    def _load_cache(self):
        if os.path.exists(self.CACHE_FILE):
            with open(self.CACHE_FILE, 'r') as f:
                try:
                    data = json.load(f)
                    self._cache = data.get("apis", {})
                    self._cache_timestamp = datetime.fromisoformat(data.get("timestamp"))
                except (json.JSONDecodeError, TypeError):
                    print(f"Warning: Could not decode JSON from {self.CACHE_FILE} or invalid cache format. Starting with empty cache.")
                    self._cache = {}
                    self._cache_timestamp = None

    def _save_cache(self):
        with open(self.CACHE_FILE, 'w') as f:
            json.dump({"apis": self._cache, "timestamp": datetime.now().isoformat()}, f, indent=2)

    async def _fetch_all_apis(self):
        client = await HTTPClientManager.get_client()
        response = await client.get(f"{self.BASE_URL}/entries")
        response.raise_for_status()
        entries = response.json()["entries"]
        self._cache = {entry["API"]: entry for entry in entries}
        self._cache_timestamp = datetime.now()
        self._save_cache()

    async def get_all_apis(self, force_refresh: bool = False) -> dict:
        if force_refresh or not self._cache or \
           (self._cache_timestamp and datetime.now() - self._cache_timestamp > timedelta(hours=self.CACHE_TTL_HOURS)):
            print("Refreshing API catalog from source...")
            await self._fetch_all_apis()
        else:
            print("Using cached API catalog.")
        return self._cache

    async def search_apis(self, query: str, category: str = None) -> list:
        apis = await self.get_all_apis()
        results = []
        query_lower = query.lower()

        for api_name, api_details in apis.items():
            if category and api_details["Category"].lower() != category.lower():
                continue

            # Simple fuzzy search: check query in API name, Description, or Category
            if query_lower in api_name.lower() or \
               query_lower in api_details["Description"].lower() or \
               query_lower in api_details["Category"].lower():
                results.append(api_details)

        # Rank by relevance (e.g., exact match, then starts with, then contains)
        def rank_api(api):
            name_lower = api["API"].lower()
            if name_lower == query_lower: return 3
            if name_lower.startswith(query_lower): return 2
            if query_lower in name_lower: return 1
            return 0

        return sorted(results, key=rank_api, reverse=True)

async def run_cli(args):
    """CLI entrypoint for querying services."""
    command = args[0] if args else "help"

    if command == "weather":
        try:
            lat = float(args[1])
            lon = float(args[2])
            temp = await WeatherService().get_current_temperature(lat, lon)
            print(f"Current temperature at ({lat}, {lon}): {temp}°C")
        except (IndexError, ValueError):
            print("Usage: weather <latitude> <longitude>")
        except httpx.HTTPStatusError as e:
            print(f"Weather service error: {e}")
        except httpx.RequestError as e:
            print(f"Weather service connection error: {e}")
    elif command == "fx":
        try:
            action = args[1]
            if action == "rates":
                base = args[2] if len(args) > 2 else "EUR"
                rates = await FXService().get_latest_rates(base_currency=base)
                print(f"Latest exchange rates (base: {base}): {json.dumps(rates, indent=2)}")
            elif action == "convert":
                amount = float(args[2])
                from_curr = args[3]
                to_curr = args[4]
                converted = await FXService().convert_currency(amount, from_curr, to_curr)
                print(f"{amount} {from_curr} = {converted:.2f} {to_curr}")
            else:
                print("Usage: fx [rates [base_currency]] | [convert <amount> <from_currency> <to_currency>]")
        except (IndexError, ValueError):
            print("Usage: fx [rates [base_currency]] | [convert <amount> <from_currency> <to_currency>]")
        except httpx.HTTPStatusError as e:
            print(f"FX service error: {e}")
        except httpx.RequestError as e:
            print(f"FX service connection error: {e}")
    elif command == "discover":
        try:
            action = args[1]
            api_discovery = APIDiscoveryService()
            if action == "all":
                apis = await api_discovery.get_all_apis(force_refresh=True)
                print(f"Found {len(apis)} public APIs.")
            elif action == "search":
                query = args[2]
                category = args[3] if len(args) > 3 else None
                results = await api_discovery.search_apis(query, category)
                print(f"Found {len(results)} APIs for query '{query}' (category: {category or 'any'}):")
                for res in results:
                    print(f"  - {res['API']}: {res['Description']} (Category: {res['Category']})")
            else:
                print("Usage: discover [all] | [search <query> [category]]")
        except (IndexError, ValueError):
            print("Usage: discover [all] | [search <query> [category]]")
        except httpx.HTTPStatusError as e:
            print(f"API Discovery service error: {e}")
        except httpx.RequestError as e:
            print(f"API Discovery service connection error: {e}")
    else:
        print("Available commands: weather, fx, discover")

    await HTTPClientManager.close_client()
````

### Triquetra Liquid Memory: Efficiency Gain Simulation

This module simulates the fine-tuning of a **Liquid Memory Substrate** based on the Triquetra archetype framework (e.g., Generate, Preserve, Dissolve).

Instead of linearly expanding the context window (Standard Memory), the Liquid Memory dynamically compresses and dissolves irrelevant context state, optimizing for semantic density and retrieval speed.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

class StandardMemory:
    """Simulates standard linear context window (grows infinitely or hard truncates)."""
    def __init__(self):
        self.context_tokens = 0
        self.processing_cost = 0

    def process_cycle(self, incoming_tokens):
        # Linearly grows, increasing processing cost quadratically (attention mechanism simulation)
        self.context_tokens += incoming_tokens
        self.processing_cost += (self.context_tokens ** 1.5) / 1000
        return self.processing_cost

class TriquetraLiquidMemory:
    """
    Simulates a dynamic cognitive substrate.
    - Generate (Incoming context)
    - Preserve (Active semantic anchoring)
    - Dissolve (Context pruning/decay)
    """
    def __init__(self, preservation_rate=0.4, dissolve_threshold=0.1):
        self.active_tokens = 0
        self.processing_cost = 0
        # Fine-tuning parameters
        self.preservation_rate = preservation_rate # How much context is compressed and kept
        self.dissolve_threshold = dissolve_threshold # Base decay rate of old context

    def fine_tune(self, new_preservation, new_dissolve):
        self.preservation_rate = new_preservation
        self.dissolve_threshold = new_dissolve

    def process_cycle(self, incoming_tokens):
        # Apply Triquetra Logic: Dissolve old, Preserve core, Generate new
        dissolved = self.active_tokens * self.dissolve_threshold
        self.active_tokens = (self.active_tokens - dissolved) * self.preservation_rate + incoming_tokens

        # Cost scales with strictly bounded active tokens, not absolute history
        self.processing_cost += (self.active_tokens ** 1.5) / 1000
        return self.processing_cost

# --- Simulation Execution ---
np.random.seed(42)
cycles = 50
incoming_streams = np.random.randint(500, 1500, cycles) # Incoming tokens per cycle

# Initialize architectures
std_memory = StandardMemory()
liquid_memory = TriquetraLiquidMemory(preservation_rate=0.8, dissolve_threshold=0.2)

# Fine-tuning the Liquid Memory for optimal density
print("Fine-tuning Triquetra Liquid Memory parameters...")
liquid_memory.fine_tune(new_preservation=0.3, new_dissolve=0.5)
print(f"Optimal Parameters -> Preservation: {liquid_memory.preservation_rate}, Dissolve: {liquid_memory.dissolve_threshold}\n")

std_costs = []
liquid_costs = []

for tokens in incoming_streams:
    std_costs.append(std_memory.process_cycle(tokens))
    liquid_costs.append(liquid_memory.process_cycle(tokens))

# Calculate Efficiency Gain
total_std_cost = sum(std_costs)
total_liquid_cost = sum(liquid_costs)
efficiency_gain = ((total_std_cost - total_liquid_cost) / total_std_cost) * 100

print(f"--- Simulation Results ({cycles} Cognitive Cycles) ---")
print(f"Total Processing Cost (Standard): {total_std_cost:,.2f} compute units")
print(f"Total Processing Cost (Liquid):   {total_liquid_cost:,.2f} compute units")
print(f"\n🔥 Triquetra Liquid Memory Efficiency Gain: {efficiency_gain:.2f}% 🔥\n")

# --- Visualization ---
plt.figure(figsize=(10, 6))
plt.plot(range(cycles), std_costs, label='Standard Memory (Linear Growth)', color='red', linestyle='--')
plt.plot(range(cycles), liquid_costs, label='Triquetra Liquid Memory (Dynamic Pruning)', color='teal', linewidth=2)
plt.fill_between(range(cycles), liquid_costs, std_costs, color='green', alpha=0.1, label='Efficiency Delta (Saved Compute)')

plt.title('Cognitive Compute Cost: Standard vs. Triquetra Liquid Memory')
plt.xlabel('Agent Cognitive Cycles (t)')
plt.ylabel('Compute Cost (Simulated O(N^1.5))')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

### Phase 2: Triquetra Liquid Memory Pipeline & LoRA Integration
This script applies the Liquid Memory concept to real-world domain text. It scores text chunks, dissolves the noise, and trains a lightweight LLM using LoRA on the highly condensed semantic data.

````python
import torch
import numpy as np
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer
from peft import LoraConfig, get_peft_model, TaskType
from datasets import Dataset

class LiquidMemoryContextProcessor:
    """
    Applies the Triquetra archetype (Preserve/Dissolve) to raw training data.
    Simulates a meta-cognitive layer that filters out boilerplate before training,
    drastically reducing the token budget and optimizing learning efficiency.
    """
    def __init__(self, preservation_rate=0.4, dissolve_threshold=0.3):
        self.preservation_rate = preservation_rate
        self.dissolve_threshold = dissolve_threshold

    def _score_semantic_density(self, text_chunk):
        # Simulated semantic density scoring.
        # In production, this would use a fast embeddings model (like SentenceTransformers)
        # or a keyword heuristic to rank the importance of the text.
        finance_keywords = ['revenue', 'acquisition', 'contract', 'predict', 'growth', 'margin', 'award', 'timing']
        score = sum([1.0 for word in finance_keywords if word in text_chunk.lower()])
        return score + np.random.uniform(0.1, 0.5) # Add base density

    def process_document(self, document: str) -> str:
        # 1. Generate: Chunk the document into processing frames (e.g., sentences/paragraphs)
        chunks = [chunk.strip() for chunk in document.split('.') if len(chunk.strip()) > 10]

        if not chunks:
            return document

        scored_chunks = [(chunk, self._score_semantic_density(chunk)) for chunk in chunks]

        # Normalize scores
        max_score = max([s for _, s in scored_chunks]) if scored_chunks else 1.0

        liquid_context = []
        for chunk, score in scored_chunks:
            normalized_score = score / (max_score + 1e-5)

            # 2. Dissolve: Discard low-density boilerplate
            if normalized_score < self.dissolve_threshold:
                continue

            # 3. Preserve: Keep high-density signals
            if normalized_score >= self.preservation_rate:
                liquid_context.append(chunk)

        # Reconstitute the highly dense context
        return "... ".join(liquid_context) + "."

# --- 1. Generate Mock Domain Dataset ---
# Domains: Business Finance, Gov Contracts, Stock Prediction, Product Timing
raw_training_data = [
    {
        "instruction": "Analyze the viability of acquiring Gov Contract ID-8842 based on financial positioning.",
        "context": "The government has issued a solicitation for infrastructure overhaul. The standard boilerplate states that all contractors must comply with section 4A. However, the core margin potential is huge. Revenue growth in this sector is up 14%. We must submit by Q3. This is standard text that doesn't mean much. The acquisition aligns with our product timing."
    },
    {
        "instruction": "Predict stock movement for TechCorp based on product positioning.",
        "context": "TechCorp announced their new AI suite. Usually, announcements are accompanied by PR fluff. We thank our partners. The predictive models show a 20% upside due to perfect market timing and aggressive product positioning against competitors. There are risks, but growth is expected. This is a disclaimer."
    }
]

# --- 2. Apply Liquid Memory Pre-processing ---
print("💧 Applying Liquid Memory Substrate to Raw Contexts...")
liquid_processor = LiquidMemoryContextProcessor(preservation_rate=0.5, dissolve_threshold=0.3)

processed_data = []
for item in raw_training_data:
    original_length = len(item['context'].split())
    condensed_context = liquid_processor.process_document(item['context'])
    new_length = len(condensed_context.split())

    print(f"-> Condensed context from {original_length} to {new_length} words. ({(1 - new_length/original_length)*100:.1f}% dissolved)")

    # Format for LLM Instruction Tuning
    prompt = f"Instruction: {item['instruction']}\nContext: {condensed_context}\nResponse:"
    processed_data.append({"text": prompt})

hf_dataset = Dataset.from_list(processed_data)

# --- 3. LoRA Model Setup ---
# Using a tiny model (e.g., gpt2) for quick demonstration in Colab.
# In production, swap 'gpt2' with 'meta-llama/Llama-2-7b-hf' or similar.
print("\n⚙️ Initializing Base Model and LoRA Adapters...")
model_id = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token

base_model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")

# Configure LoRA
lora_config = LoraConfig(
    r=8,
    lora_alpha=32,
    target_modules=["c_attn"], # Target attention blocks in GPT-2
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.CAUSAL_LM
)

peft_model = get_peft_model(base_model, lora_config)
peft_model.print_trainable_parameters()

# Tokenize the condensed dataset
def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)

tokenized_datasets = hf_dataset.map(tokenize_function, batched=True)

# --- 4. Training Loop ---
print("\n🚀 Commencing Fine-Tuning on Liquid Memory Dataset...")
training_args = TrainingArguments(
    output_dir="./lora-liquid-memory-finance",
    learning_rate=2e-4,
    per_device_train_batch_size=1,
    num_train_epochs=3,
    logging_steps=1,
    remove_unused_columns=False,
    report_to="none"
)

trainer = Trainer(
    model=peft_model,
    args=training_args,
    train_dataset=tokenized_datasets,
)

trainer.train()
print("✅ Training Complete. The model has internalized the concentrated semantic signals.")
````

## One Mind Architecture: Mindprint & Hybrid Retrieval

This framework implements the requested architecture:
1. **Dewey Decimal Nanochemistry Taxonomy**: A hierarchical naming convention for latent states.
2. **Hybrid Retrieval (Mindprint)**: Combines Keyword (BM25) and Semantic (Dense Embedding) retrieval to extract the 'Mindprint' context.
3. **SLM Integration**: Trains a Small Language Model conditioned on these precisely retrieved latent states.

````python
!pip install -q transformers datasets sentence-transformers rank_bm25
````

````python
import numpy as np
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer

# 1. Dewey Decimal Nanochemistry Naming Convention
class DeweyNanochemTaxonomy:
    def __init__(self):
        self.taxonomy = {
            "541.39.1": "Latent Bond Activation (Catalysis)",
            "541.39.2": "Fullerene Cognitive Routing (Carbon Structures)",
            "541.39.3": "Quantum Dot Memory Substrates",
            "541.39.4": "Plasmonic Semantic Resonance"
        }

    def map_to_taxonomy(self, embedding_vector):
        # In a real system, this would map the latent vector space to the taxonomy.
        # Simulating deterministic routing based on vector sum for demonstration.
        val = np.sum(embedding_vector) % 4
        keys = list(self.taxonomy.keys())
        return keys[int(val)]

# 2. Hybrid Retrieval Engine (Mindprint)
class HybridMindprintRetriever:
    def __init__(self, corpus):
        self.corpus = corpus
        self.taxonomy = DeweyNanochemTaxonomy()

        # Semantic Setup
        print("Initializing Semantic Latent Space (SentenceTransformer)...")
        self.encoder = SentenceTransformer('all-MiniLM-L6-v2')
        self.embeddings = self.encoder.encode(corpus)

        # Keyword Setup
        print("Initializing Keyword Lexical Space (BM25)...")
        tokenized_corpus = [doc.split(" ") for doc in corpus]
        self.bm25 = BM25Okapi(tokenized_corpus)

    def retrieve_mindprint(self, query, alpha=0.5, top_k=2):
        # Semantic Score (Cosine Similarity)
        query_emb = self.encoder.encode([query])[0]
        semantic_scores = np.dot(self.embeddings, query_emb) / (np.linalg.norm(self.embeddings, axis=1) * np.linalg.norm(query_emb))

        # Keyword Score
        tokenized_query = query.split(" ")
        keyword_scores = self.bm25.get_scores(tokenized_query)

        # Normalize Keyword Scores to 0-1
        if np.max(keyword_scores) > 0:
            keyword_scores = keyword_scores / np.max(keyword_scores)

        # Hybrid Score
        hybrid_scores = (alpha * semantic_scores) + ((1 - alpha) * keyword_scores)

        # Top K Indices
        top_indices = np.argsort(hybrid_scores)[::-1][:top_k]

        results = []
        for idx in top_indices:
            taxonomy_code = self.taxonomy.map_to_taxonomy(self.embeddings[idx])
            results.append({
                "text": self.corpus[idx],
                "hybrid_score": hybrid_scores[idx],
                "nanochem_classification": f"{taxonomy_code} : {self.taxonomy.taxonomy[taxonomy_code]}"
            })
        return results

# Simulated Knowledge Base
knowledge_corpus = [
    "Carbon nanotubes enhance neural network routing efficiency by 40%.",
    "Quantum dots act as liquid memory substrates for rapid state recall.",
    "Plasmonic resonance enables zero-latency semantic mapping in SLMs.",
    "Latent catalytic bonds trigger rapid gradient descent convergence."
]

mindprint_retriever = HybridMindprintRetriever(knowledge_corpus)

print("\n--- Testing Hybrid Mindprint Retrieval ---")
query = "liquid memory routing"
retrieved_states = mindprint_retriever.retrieve_mindprint(query)
for i, state in enumerate(retrieved_states):
    print(f"\nState {i+1}:")
    print(f"Classification: {state['nanochem_classification']}")
    print(f"Content: {state['text']}")
    print(f"Score: {state['hybrid_score']:.4f}")
````

### SLM Training Loop (The 'One Mind' Integration)

We now define the training pipeline that takes the hybrid retrieved `Mindprint` (categorized by our Nanochemistry Dewey system) and formats it into the context window for the SLM. This forces the model to ground its generative space in the structured latent taxonomy.

````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer
from datasets import Dataset

# 3. Model & Tokenizer Setup (Using a lightweight SLM for demonstration)
model_id = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token
model = AutoModelForCausalLM.from_pretrained(model_id)

# Generate Training Data fused with Mindprints
training_queries = [
    "How do we improve memory substrates?",
    "What activates the semantic mapping?",
    "Explain routing efficiency."
]

# Construct 'One Mind' Prompts
training_data = []
for q in training_queries:
    # Retrieve Mindprint context
    states = mindprint_retriever.retrieve_mindprint(q, top_k=1)
    latent_context = states[0]['text']
    dewey_code = states[0]['nanochem_classification']

    # The prompt explicitly encodes the taxonomy and latent state
    fused_prompt = f"[MINDPRINT STATE: {dewey_code}]\n[CONTEXT: {latent_context}]\nUser: {q}\nSystem: Based on the latent {dewey_code.split(':')[0].strip()} architecture, "
    training_data.append({"text": fused_prompt})

dataset = Dataset.from_list(training_data)

def tokenize_func(examples):
    tokenized = tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)
    tokenized["labels"] = tokenized["input_ids"].copy() # Explicitly add labels for loss calculation
    return tokenized

tokenized_dataset = dataset.map(tokenize_func, batched=True)

# 4. SLM Training Config
training_args = TrainingArguments(
    output_dir="./one_mind_slm",
    per_device_train_batch_size=2,
    num_train_epochs=3,
    learning_rate=3e-4,
    logging_steps=1,
    report_to="none",
    dataloader_pin_memory=False # Set to False as no accelerator is found
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
)

print("\n--- Initiating 'One Mind' SLM Training ---")
trainer.train()
print("Training Complete. The model is now conditioned on the Mindprint Hybrid Latent Taxonomy.")

# 5. Model Evaluation (Validation Step)
print("\n--- Evaluating 'One Mind' SLM ---")
model.eval()  # Set model to evaluation mode

with torch.no_grad():
    # Use the same tokenized_dataset for simplicity as a validation set
    eval_dataloader = trainer.get_eval_dataloader(tokenized_dataset)
    total_eval_loss = 0
    for batch in eval_dataloader:
        # Move batch to the same device as the model
        batch = {k: v.to(model.device) for k, v in batch.items()}
        outputs = model(**batch)
        loss = outputs.loss
        total_eval_loss += loss.item()

    avg_eval_loss = total_eval_loss / len(eval_dataloader)
    print(f"Average Validation Loss: {avg_eval_loss:.4f}")

# 6. Save the fine-tuned model weights
torch.save(model.state_dict(), "one_mind_slm_model.pth")
print("Fine-tuned 'One Mind' SLM model weights saved to 'one_mind_slm_model.pth'.")
````

````python
# Relationship between df_novel_concepts and df_ip_framework:
# Conceptually in this ecosystem, `df_novel_concepts` acts as a repository for raw,
# unstructured market gaps, ideations, and architectures (like the biomimetic designs).
# `df_ip_framework` structures these concepts into formal, defensible intellectual property
# assets and taxonomic mappings. The relationship is a direct pipeline: raw ideation
# flows from novel_concepts and is engineered into structured assets in the ip_framework.

import re

print("Filling missing writer values...")
tweets_df['writer'] = tweets_df['writer'].fillna('Unknown')

print("Filtering for high-engagement tweets (likes >= 50, retweets >= 10)...")
high_eng_df = tweets_df[(tweets_df['like_num'] >= 50) & (tweets_df['retweet_num'] >= 10)].copy()

print("Cleaning text bodies...")
def clean_text(text):
    text = re.sub(r'http\S+', '', text)
    return re.sub(r'\s+', ' ', text).strip()

high_eng_df['body'] = high_eng_df['body'].astype(str).apply(clean_text)

# Sample down for quick processing during the LoRA fine-tuning step
sampled_tweets_df = high_eng_df.sample(n=min(2000, len(high_eng_df)), random_state=42)
print(f"Preprocessing complete. Final sampled dataset shape: {sampled_tweets_df.shape}")
display(sampled_tweets_df[['writer', 'like_num', 'retweet_num', 'body']].head(3))
````

````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer
from peft import LoraConfig, get_peft_model, TaskType

model_id = "gpt2"

print(f"Loading tokenizer and model for {model_id}...")
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token
base_model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")

print("Configuring LoRA...")
lora_config = LoraConfig(
    r=8,
    lora_alpha=32,
    target_modules=["c_attn"],
    lora_dropout=0.05,
    bias="none",
    task_type=TaskType.CAUSAL_LM
)

peft_model = get_peft_model(base_model, lora_config)
peft_model.print_trainable_parameters()

print("Tokenizing dataset...")
def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)

tokenized_datasets = tweet_dataset.map(tokenize_function, batched=True)

print("Starting LoRA fine-tuning...")
training_args = TrainingArguments(
    output_dir="./lora-twitter-finetuned",
    learning_rate=2e-4,
    per_device_train_batch_size=2,
    num_train_epochs=1, # Keep small for demonstration
    logging_steps=50,
    remove_unused_columns=False,
    report_to="none"
)

trainer = Trainer(
    model=peft_model,
    args=training_args,
    train_dataset=tokenized_datasets,
)

trainer.train()
print("Fine-tuning complete.")

torch.save(peft_model.state_dict(), "lora_twitter_model.pth")
print("Model saved to lora_twitter_model.pth")
````
